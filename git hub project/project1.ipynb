```python
import random as r
print('Game rules\n'+
      'enter 1 for rock\n'+
      'enter 2 for paper\n'+
      'enter 3 for sisscors')
while True:
    while True:
            try:
                user_choice=int(input('enter a choice'))
                break
            except Exception as e:
                print('Plz enter correct input according to the rules')

    if user_choice==1:
        user_choice='rock'
    elif user_choice==2:
        user_choice='paper'
    elif user_choice==3:
        user_choice='sisscors'

    comp_choice=r.choice([1,2,3])

    if comp_choice==1:
        comp_choice='rock'
    elif comp_choice==2:
        comp_choice='paper'
    elif comp_choice==3:
        comp_choice='sisscors'


    print(user_choice)

    print(comp_choice)


    if user_choice==comp_choice:
        print('draw')
    elif user_choice=='rock' and comp_choice=='sisscors':
        print('u wins')
    elif user_choice=='sisscors' and comp_choice=='paper':
        print('u wins')
    elif user_choice=='paper' and comp_choice=='rock':
        print('u wins')
    elif comp_choice=='rock' and user_choice=='sisscors':
        print('computer wins')
    elif comp_choice=='sisscors' and user_choice=='paper':
        print('computer wins')
    elif comp_choice=='paper' and user_choice=='rock':
        print('computer wins')
    t=input('u want to continue?y/n')
    if t.upper() =='N':
        break

```
